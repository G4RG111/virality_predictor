# 01 — iHUT PDF Exploration

Explore the 3 SharkNinja Chanel iHUT PDFs.

Goals:
- Understand slide structure and text density
- Validate verbatim extraction quality
- Identify pre-launch hackability signals
- Count chunk types per PDF

```python
import sys
sys.path.insert(0, '../')

from pathlib import Path
from pipeline.ingestion.ihut_extractor import extract_ihut_pdf, extract_multiple_ihut_pdfs

# Path to raw iHUT PDFs (placed in the parent project dir)
IHUT_DIR = Path('../../')
pdf_files = list(IHUT_DIR.glob('*.pdf'))
print(f'Found {len(pdf_files)} PDF files:')
for f in pdf_files:
    print(f'  {f.name}')
```

```python
# Extract first PDF
if pdf_files:
    doc = extract_ihut_pdf(pdf_files[0])
    print(f'Product: {doc.product}')
    print(f'Market:  {doc.market}')
    print(f'Chunks:  {len(doc.chunks)} total')
    print(f'  Verbatims: {len(doc.verbatim_chunks)}')
    print(f'  Hackability signals: {len(doc.hackability_chunks)}')
```

```python
# Sample verbatim quotes
print('--- Sample verbatim quotes ---')
for chunk in doc.verbatim_chunks[:10]:
    print(f'[Slide {chunk.slide_number}] {chunk.text[:200]}')
    print()
```

```python
# Pre-launch hackability signals
print('--- Pre-launch hackability signals ---')
for chunk in doc.hackability_chunks:
    print(f'[Slide {chunk.slide_number}] {chunk.text[:300]}')
    print()
```

```python
# Process all PDFs
all_docs = extract_multiple_ihut_pdfs(pdf_files)
print('\nSummary across all iHUT PDFs:')
for doc in all_docs:
    print(f'  {doc.product} ({doc.market}): {len(doc.chunks)} chunks, {len(doc.verbatim_chunks)} verbatims, {len(doc.hackability_chunks)} hack signals')
```
